## Objects and classes

Object-oriented programming is based on the idea of grouping data and accompanying functionality in a template, called a **class**. A class can be seen as a complex data type. Variables that are of that data type are said to be **objects** or **instances** of that class.

Let's explore the example of the `Dog` class again:

```python
class Dog:
    def __init__(self, breed, name):
        self.breed = breed
        self.name = name
```

We defined a class `Dog`, which contains a special function `__init__`. When Python functions start and end with double underscores, this generally means they are special functions that are used internally, and that are not meant to be called directly. Indeed, the initialization function `__init__` (called a **constructor**) that we defined will be called internally when an object of the specified class is initialized. Let's do so:

```python
lassie = Dog('collie', 'Lassie')
print(lassie.name + ' is a ' + lassie.breed)
```

Functions within a class are called **methods**. The constructor assigns the two parameters to variables that *belong to the object*. Within a class definition the object is always represented by `self`.

The first argument of a method is always `self`, and it will always point to the instance of the class. This first argument however is never explicitly specified when you call the method. It is implicitly passed by Python itself. That is why you see a discrepancy between the number of arguments in the instantiation and in the class definition.

Any variable or method in a class can be accessed using the period (`.`) syntax:

    object.attribute

or:

    object.method

In the above example we printed the name and breed. We can turn this into a method as well, such that each dog can be introduced. Let's extend our example:

```python
class Dog:
    def __init__(self, breed, name):
        self.breed = breed
        self.name = name

    def introduce(self):
        print(self.name + ' is a ' + self.breed)

lassie = Dog('collie', 'Lassie')
lassie.introduce()
```

Do you see what happens here? Do you understand the role of `self` and the notation with the period?

As we pointed out in the lecture, we have already made use of objects and methods throughout this course. Things like strings, lists, sets and dictionaries are all objects; the object-oriented paradigm is ubiquitous in Python.

## Inheritance

An interesting feature of object-oriented programming is that you can build more specialized classes on top of more generic classes. `Dog` for instance is a rather generic concept. We can use this generic class to build a more specialised class `K9`, a police dog with special abilities. If you use inheritance, everything that the parent class could do, the inherited class can do as well.

The syntax for inheritance is as follows, do not confuse it with parameters in a function/method definition. We also add an extra method `search_explosives`; otherwise `K9` would be no different from `Dog`:

```python
class Dog:
    def __init__(self, breed, name):
        self.breed = breed
        self.name = name

    def introduce(self):
        print(self.name + ' is a ' + self.breed)


class K9(Dog): # this class inherits the class above
    def search_explosives(self):
        print('Searching..')

max_pd = K9('shepherd', 'Max')
max_pd.introduce()
max_pd.search_explosives()
```

If we create a new method `introduce()` in the subclass `K9`, then it overrides the one in the `Dog` class:

```python
class K9(Dog):
    def introduce(self):
        print(self.name + ' is a ' + self.breed + ', and works as a police dog')

    def search_explosives(self):
        print('Searching..')

max_pd = K9('shepherd', 'Max')
max_pd.introduce()
```

Instead of completely overriding a method, you can also call the method of the parent class. The following example contains a modified version of the `introduce()` method for the `K9` class; the edited method calls the method of the parent class to avoid repetition of code, and then adds some additional functionality. In order to call the `introduce()` method of the parent class, we use the built-in function `super()`:

```python
class K9(Dog):
    def introduce(self):
        super().introduce()
        print(self.name + ' works as a police dog')

    def search_explosives(self):
        print('Searching..')

max_pd = K9('shepherd', 'Max')
max_pd.introduce()
```
